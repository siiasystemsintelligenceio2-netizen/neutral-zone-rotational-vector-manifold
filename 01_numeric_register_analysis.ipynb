# NZ-RVM + NRL-01 Register Analysis

This notebook loads the handwritten register CSV and computes digit features.

```python
from pathlib import Path
import sys, csv, json
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))
from nzrvm_math.registers import analyze_register
raw = ROOT / 'data' / 'raw' / 'handwritten-registers.csv'
print(raw)
```

```python
with raw.open(newline='', encoding='utf-8') as f:
    rows = list(csv.DictReader(f))
register_rows = [r for r in rows if 'register' in r['type'] or r['type'] == 'multiplicity']
for r in register_rows:
    features = analyze_register(r['raw_text'])
    print(r['page_id'], r['line_id'], r['raw_text'], 'length=', features.length, 'sum=', features.digit_sum, 'm6=', features.multiplicity_6)
```
